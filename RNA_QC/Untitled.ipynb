```python
def is_outlier(adata, metric: str, nmads: int):
    M = adata.obs[metric]
    #outlier = (M < np.median(M) - nmads * median_abs_deviation(M)) | (
    #    np.median(M) + nmads * median_abs_deviation(M) < M
    #)
    #return outlier

    med = np.median(M)
    mad = median_abs_deviation(M)

    lower = med - nmads * mad
    upper = med + nmads * mad

    return lower, upper
```
